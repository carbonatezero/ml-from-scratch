# Dropout
Dropout [1] is a technique for regularizing neural networks by randomly setting some output activations to zero during the forward pass. In this exercise, you will implement a dropout layer and modify your fully connected network to optionally use dropout.

[1] [Geoffrey E. Hinton et al, "Improving neural networks by preventing co-adaptation of feature detectors", arXiv 2012](https://arxiv.org/abs/1207.0580)

## Inline Question 1:
What happens if we do not divide the values being passed through inverse dropout by `p` in the dropout layer? Why does that happen?

## Answer:
If we do not divide the kept activations by `p`, then the expected output of the dropout layer will be smaller during training. Since each activation is kept only with probability `p`, its expected value becomes `p * x` instead of `x`. This changes the scale of the activations seen by later layers during training, but at test time dropout is turned off and the activations have their full scale again. Dividing by `p` in inverted dropout keeps the expected activation size the same in train and test modes.

In the file `../src/classifiers/fc_net.py`, modify your implementation to use dropout. # Regularization Experiment
As an experiment, we will train a pair of two-layer networks on 500 training examples: one will use no dropout, and one will use a keep probability of 0.25. We will then visualize the training and validation accuracies of the two networks over time.

```python
# Train two identical nets, one with dropout and one without.
np.random.seed(231)
num_train = 500
small_data = {
    'X_train': data['X_train'][:num_train],
    'y_train': data['y_train'][:num_train],
    'X_val': data['X_val'],
    'y_val': data['y_val'],
}

solvers = {}
dropout_choices = [1, 0.25]
for dropout_keep_ratio in dropout_choices:
    model = FullyConnectedNet(
        [500],
        dropout_keep_ratio=dropout_keep_ratio
    )
    print(dropout_keep_ratio)

    solver = Solver(
        model,
        small_data,
        num_epochs=25,
        batch_size=100,
        update_rule='adam',
        optim_config={'learning_rate': 5e-4,},
        verbose=True,
        print_every=100
    )
    solver.train()
    solvers[dropout_keep_ratio] = solver
    print()
```

```python
# Plot train and validation accuracies of the two models.
fig, axes = plt.subplots(2, 1, figsize=(10, 9))

for dropout_keep_ratio in dropout_choices:
    axes[0].plot(
        solvers[dropout_keep_ratio].train_acc_history, 'o-', label='%.2f dropout_keep_ratio' % dropout_keep_ratio)
axes[0].set_title('Train accuracy', pad=12)
axes[0].set_xlabel('Epoch', labelpad=8)
axes[0].set_ylabel('Accuracy')
axes[0].legend(ncol=2, loc='lower right')
axes[0].grid(True, alpha=0.3)

for dropout_keep_ratio in dropout_choices:
    axes[1].plot(
        solvers[dropout_keep_ratio].val_acc_history, 'o-', label='%.2f dropout_keep_ratio' % dropout_keep_ratio)
axes[1].set_title('Validation accuracy', pad=12)
axes[1].set_xlabel('Epoch', labelpad=8)
axes[1].set_ylabel('Accuracy')
axes[1].legend(ncol=2, loc='lower right')
axes[1].grid(True, alpha=0.3)

fig.tight_layout(h_pad=2.5)
plt.show()
```

## Inline Question 2:
Compare the validation and training accuracies with and without dropout -- what do your results suggest about dropout as a regularizer?

## Answer:
The model without dropout usually gets higher training accuracy, since it can fit the small training set more easily. The model with dropout often has lower training accuracy, but its validation accuracy can be comparable or better because dropout makes the network rely on more distributed features instead of memorizing the training examples. This suggests that dropout acts as a regularizer: it reduces overfitting by narrowing the gap between training and validation performance.
