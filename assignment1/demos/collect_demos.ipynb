# Collect Demos - Zip + Combined PDF

Run this notebook after the demo notebooks are complete.

It creates:

- `demos_code_bundle.zip` with the numbered demo notebooks and `src/` code
- `demos_combined.pdf` with the numbered demo notebooks merged in order

The combined notebook order is:

1. `01_knn_demo_repo.ipynb`
2. `02_softmax_repo.ipynb`
3. `03_two_layer_net_repo.ipynb`
4. `04_features_repo.ipynb`
5. `05_fully_connected_nets_repo.ipynb`

```python
from pathlib import Path
import os
import subprocess
import sys

demo_dir = Path.cwd()
if demo_dir.name != "demos":
    candidates = [demo_dir / "assignment1" / "demos", demo_dir / "demos"]
    demo_dir = next(path for path in candidates if (path / "collect_demos.sh").exists())

env = os.environ.copy()
env["PYTHON"] = sys.executable

subprocess.run(["bash", "collect_demos.sh"], cwd=demo_dir, env=env, check=True)
```
